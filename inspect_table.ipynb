## Imports and Preamble

```python
from pathlib import Path
from astropy import table
import numpy as np
```

```python
READ_DIR = Path(
    'data/tables/opt_comp_stars/TOI-431/crit999_mandel4/unbin_results_all.fits'
)

t = table.Table.read(READ_DIR)
```

```python
NIGHT = 20231007
BAND = 'V'

selected_t = [(t['band'] == BAND) & (t['night'] == NIGHT)]
selected_t.sort('jd')

print(f'Number of selected exposures: {len(selected_t)}')
```

## Table Inspection

```python
for n in np.unique(t['night']):
    print(
        [len(t[(t['night'] == n) & (t['band'] == b)]) for b in ['B', 'V', 'R', 'I']]
    )
```
