```python
from astropy.table import QTable
import astropy.units as u
from astropy.time import Time
import matplotlib.pyplot as plt
```

```python
t = QTable.read(
    'data/tables/opt_comp_stars/TOI-431/'
    'toi431_gaia_epoch_2026-10-24-05-28-34-009393.fits'
)
```

```python
t
```

```python
# Reference epoch
epoch_ref = Time('2010-01-01T00:00:00', scale='tcb')  # or format='isot'
# Add elapsed days and convert to datetime objects
times_dt = (epoch_ref + t['g_transit_time'][0].value * u.day).to_datetime()
```

```python
flux = t['g_transit_flux'][0].value
err = t['g_transit_flux_error'][0].value
norm = flux.mean()

mask = t['g_transit_flux_over_error'][0] > 1000
```

```python
plt.figure(figsize=(6,3))
plt.errorbar(
    times_dt[mask], flux[mask] / norm,
    yerr=err[mask] / norm, fmt='o'
)
```

```python
plt.errorbar(
    times_dt,
    t['g_transit_flux_over_error'][0],
    fmt='o'
)
```
